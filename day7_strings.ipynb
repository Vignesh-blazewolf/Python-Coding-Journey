```python
# 📅 Day 7: Strings Practice

# 1. Create a string
text = "Hello World"
print(text)

# 2. String indexing
print("First character:", text[0])

# 3. String slicing
print("Slice:", text[0:5])   # Hello

# 4. Reverse a string
print("Reversed:", text[::-1])

# 5. String length
print("Length:", len(text))

# 6. Convert case
print(text.upper())
print(text.lower())

# 7. Remove spaces
name = "  Vignesh  "
print(name.strip())

# 8. Replace words
print(text.replace("World", "Python"))

# 9. Split string
sentence = "Python is easy to learn"
words = sentence.split()
print(words)

# 10. Join strings
joined = "-".join(words)
print(joined)

# 11. Check palindrome
word = input("Enter a word: ")
if word == word[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")

# 12. Count vowels
text = input("Enter a string: ")
vowels = "aeiouAEIOU"
count = 0
for char in text:
    if char in vowels:
        count += 1
print("Vowel count:", count)
```
