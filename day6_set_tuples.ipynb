```python
# 📅 Day 21: Sets & Tuples Practice

# ----------- TUPLES -----------

# 1. Create a tuple
t = (1, 2, 3, 4)
print("Tuple:", t)

# 2. Access tuple elements
print("First element:", t[0])

# 3. Tuple is immutable (cannot change)
# t[0] = 10  ❌ This will give error

# 4. Tuple unpacking
a, b, c = (10, 20, 30)
print(a, b, c)

# 5. Count and index
nums = (1, 2, 2, 3)
print("Count of 2:", nums.count(2))
print("Index of 3:", nums.index(3))

# ----------- SETS -----------

# 6. Create a set
s = {1, 2, 3, 4}
print("Set:", s)

# 7. Duplicate removal
dup = {1, 2, 2, 3, 3}
print("Unique values:", dup)

# 8. Add element
s.add(5)
print("After add:", s)

# 9. Remove element
s.remove(2)
print("After remove:", s)

# 10. Union of sets
a = {1, 2, 3}
b = {3, 4, 5}
print("Union:", a.union(b))

# 11. Intersection
print("Intersection:", a.intersection(b))

# 12. Difference
print("Difference:", a.difference(b))
```
