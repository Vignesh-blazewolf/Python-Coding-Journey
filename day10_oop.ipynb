```python
# 📅 Day 10: OOP Practice

# 1. Create a class
class Student:
    pass

s1 = Student()
print("Object created:", s1)

# 2. Class with attributes
class Student:
    name = "Vignesh"
    age = 21

s2 = Student()
print(s2.name, s2.age)

# 3. Constructor (__init__)
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

s3 = Student("Arun", 20)
print(s3.name, s3.age)

# 4. Instance method
class Student:
    def __init__(self, name):
        self.name = name

    def greet(self):
        print("Hello, I am", self.name)

s4 = Student("Kumar")
s4.greet()

# 5. Multiple objects
s1 = Student("A")
s2 = Student("B")
print(s1.name, s2.name)

# 6. Update object property
s1.age = 22
print(s1.age)

# 7. Delete property
del s1.age

# 8. Delete object
del s2

# 9. Class with multiple methods
class Calculator:
    def add(self, a, b):
        return a + b

    def multiply(self, a, b):
        return a * b

calc = Calculator()
print(calc.add(2, 3))
print(calc.multiply(2, 3))

# 10. Encapsulation (private variable)
class Person:
    def __init__(self, name):
        self.__name = name  # private

    def get_name(self):
        return self.__name

p = Person("Vignesh")
print(p.get_name())

# 11. Simple bank system
class Bank:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
        else:
            print("Insufficient balance")

acc = Bank(1000)
acc.deposit(500)
acc.withdraw(200)
print(acc.balance)

# 12. String representation
class Student:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return f"Student: {self.name}"

s = Student("Vignesh")
print(s)
```
