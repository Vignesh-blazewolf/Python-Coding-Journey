```python
# 📅 Day 9: Exception Handling Practice

# 1. Basic try-except
try:
    num = int(input("Enter a number: "))
    print(num)
except:
    print("Invalid input")

# 2. Specific exception (ValueError)
try:
    num = int(input("Enter number: "))
except ValueError:
    print("Please enter a valid integer")

# 3. Division by zero
try:
    a = int(input("Enter number: "))
    b = int(input("Enter number: "))
    print(a / b)
except ZeroDivisionError:
    print("Cannot divide by zero")

# 4. Multiple exceptions
try:
    x = int(input("Enter number: "))
    y = int(input("Enter number: "))
    print(x / y)
except ValueError:
    print("Invalid input")
except ZeroDivisionError:
    print("Division by zero error")

# 5. Using else
try:
    num = int(input("Enter number: "))
except ValueError:
    print("Error occurred")
else:
    print("You entered:", num)

# 6. Using finally
try:
    f = open("sample.txt", "r")
    print(f.read())
except FileNotFoundError:
    print("File not found")
finally:
    print("Execution completed")

# 7. Raise custom exception
age = int(input("Enter age: "))
if age < 18:
    raise Exception("You must be 18+")

# 8. Custom validation
try:
    password = input("Enter password: ")
    if len(password) < 6:
        raise ValueError("Password too short")
except ValueError as e:
    print(e)

# 9. Index error
try:
    lst = [1, 2, 3]
    print(lst[5])
except IndexError:
    print("Index out of range")

# 10. Key error
try:
    data = {"name": "Vignesh"}
    print(data["age"])
except KeyError:
    print("Key not found")

# 11. Type error
try:
    print("Hello" + 5)
except TypeError:
    print("Type mismatch")

# 12. Generic exception handling
try:
    x = int("abc")
except Exception as e:
    print("Error:", e)
```
