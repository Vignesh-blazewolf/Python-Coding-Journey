```python
# 📅 Day 18: Functions Practice

# 1. Simple function
def greet():
    print("Hello, Welcome!")

greet()

# 2. Function with parameter
def greet_user(name):
    print("Hello", name)

greet_user("Vignesh")

# 3. Function with return value
def add(a, b):
    return a + b

result = add(5, 10)
print("Sum:", result)

# 4. Even or Odd using function
def check_even(num):
    if num % 2 == 0:
        return "Even"
    else:
        return "Odd"

print(check_even(7))

# 5. Largest of two numbers
def largest(a, b):
    return a if a > b else b

print("Largest:", largest(10, 20))

# 6. Factorial using function
def factorial(n):
    fact = 1
    for i in range(1, n + 1):
        fact *= i
    return fact

print("Factorial:", factorial(5))

# 7. Prime check function
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True

print("Prime:", is_prime(7))

# 8. Function with default parameter
def greet_default(name="User"):
    print("Hello", name)

greet_default()
greet_default("Vignesh")

# 9. Multiple return values
def calc(a, b):
    return a + b, a - b, a * b

sum_val, sub_val, mul_val = calc(10, 5)
print(sum_val, sub_val, mul_val)

# 10. Lambda function
square = lambda x: x * x
print("Square:", square(5))

# 11. Function to count vowels
def count_vowels(text):
    vowels = "aeiouAEIOU"
    count = 0
    for char in text:
        if char in vowels:
            count += 1
    return count

print("Vowels:", count_vowels("Hello World"))

# 12. Recursive function (factorial)
def factorial_rec(n):
    if n == 1:
        return 1
    return n * factorial_rec(n - 1)

print("Recursive Factorial:", factorial_rec(5))
```
