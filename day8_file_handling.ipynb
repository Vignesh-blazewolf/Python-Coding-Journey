```python
# 📅 Day 23: File Handling Practice

# 1. Write to a file
file = open("sample.txt", "w")
file.write("Hello, this is Day 23\n")
file.close()

# 2. Read file
file = open("sample.txt", "r")
print(file.read())
file.close()

# 3. Append to file
file = open("sample.txt", "a")
file.write("Appending new line\n")
file.close()

# 4. Read line by line
file = open("sample.txt", "r")
for line in file:
    print(line.strip())
file.close()

# 5. Using with (best practice)
with open("sample.txt", "r") as file:
    content = file.read()
    print(content)

# 6. Count lines
with open("sample.txt", "r") as file:
    count = len(file.readlines())
    print("Lines:", count)

# 7. Count words
with open("sample.txt", "r") as file:
    words = file.read().split()
    print("Words:", len(words))

# 8. Copy file content
with open("sample.txt", "r") as src:
    with open("copy.txt", "w") as dest:
        dest.write(src.read())

# 9. Write user input to file
data = input("Enter text: ")
with open("user.txt", "w") as file:
    file.write(data)

# 10. Read specific lines
with open("sample.txt", "r") as file:
    lines = file.readlines()
    print(lines[:2])

# 11. Check if file exists
import os
if os.path.exists("sample.txt"):
    print("File exists")
else:
    print("File not found")

# 12. Delete a file
import os
if os.path.exists("copy.txt"):
    os.remove("copy.txt")
    print("File deleted")
```
